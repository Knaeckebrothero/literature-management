```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


# Load the data
df = pd.read_csv('content_matrix.csv')

# Clean up by removing empty rows and columns
df = df.iloc[:10, :24]  # Only take the first 10 rows and 24 columns

# For binary questions (0/1), we'll create a heatmap
binary_questions = [
    "Does the system use Retrieval-Augmented Generation (RAG)?",
    "Does the system support multimodal content?",
    "Is student learning motivation explicitly addressed in the design?",
    "Assessment capabilities?",
    "Empirical evidence provided?"
]

# For categorical questions, we'll create bar charts
categorical_questions = [
    "Which language model is implemented?",
    "Primary target audience of the virtual tutor?",
    "Academic disciplines supported by the tutor?",
    "Integration with learning management systems?",
    "Implementation status?"
]

# Create a figure for the heatmap of binary questions
plt.figure(figsize=(20, 6))
binary_data = df[df['Questions'].isin(binary_questions)].copy()
binary_data = binary_data.set_index('Questions')
binary_data = binary_data.astype(float)  # Convert to numeric

# Create heatmap
sns.heatmap(binary_data, cmap="YlGnBu", cbar_kws={'label': 'Feature Present'},
            linewidths=.5, linecolor='gray',
            annot=True, fmt=".0f", annot_kws={"size": 10})
plt.title('Binary Features Across Papers', fontsize=16)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.savefig('binary_features_heatmap.png', dpi=300, bbox_inches='tight')
plt.close()

# For categorical questions, let's generate frequency counts for the most common values
plt.figure(figsize=(20, 15))

for i, question in enumerate(categorical_questions):
    plt.subplot(3, 2, i+1)

    # Extract row data for this question
    row_data = df[df['Questions'] == question].iloc[0, 1:].value_counts()

    # Create bar chart
    colors = plt.cm.viridis(np.linspace(0, 0.9, len(row_data)))
    row_data.plot(kind='bar', color=colors)

    plt.title(question, fontsize=12)
    plt.ylabel('Number of Papers', fontsize=10)
    plt.xlabel('Categories', fontsize=10)
    plt.xticks(rotation=45, ha='right', fontsize=9)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('categorical_features_barcharts.png', dpi=300, bbox_inches='tight')
plt.close()

# Create a comprehensive visual summary - simpler version focusing on key insights
plt.figure(figsize=(20, 10))

# Summarize language models
plt.subplot(2, 2, 1)
lm_data = df[df['Questions'] == "Which language model is implemented?"].iloc[0, 1:].value_counts()
lm_data.plot(kind='pie', autopct='%1.1f%%', colors=plt.cm.tab10.colors, startangle=90)
plt.title('Language Models Used', fontsize=14)
plt.ylabel('')

# Summarize implementation status
plt.subplot(2, 2, 2)
status_data = df[df['Questions'] == "Implementation status?"].iloc[0, 1:].value_counts()
status_data.plot(kind='pie', autopct='%1.1f%%', colors=plt.cm.Paired.colors, startangle=90)
plt.title('Implementation Status', fontsize=14)
plt.ylabel('')

# Summarize academic disciplines
plt.subplot(2, 2, 3)
discipline_data = df[df['Questions'] == "Academic disciplines supported by the tutor?"].iloc[0, 1:].value_counts()
discipline_data.plot(kind='bar', color=plt.cm.Set3.colors)
plt.title('Academic Disciplines', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Count of special features
plt.subplot(2, 2, 4)
features = []
counts = []

for q in binary_questions:
    feature_count = df[df['Questions'] == q].iloc[0, 1:].astype(float).sum()
    features.append(q.replace('Does the system use ', '').replace('Is student ', '').replace('explicitly addressed in the design?', '').replace('?', ''))
    counts.append(feature_count)

feature_df = pd.Series(counts, index=features)
feature_df.plot(kind='bar', color=plt.cm.tab20c.colors)
plt.title('Feature Adoption (out of 23 papers)', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylabel('Number of Papers')

plt.tight_layout()
plt.savefig('summary_visualization.png', dpi=300, bbox_inches='tight')
plt.close()

print("Visualizations generated:")
print("1. binary_features_heatmap.png - Heatmap showing binary features across papers")
print("2. categorical_features_barcharts.png - Bar charts showing distribution of categorical features")
print("3. summary_visualization.png - Summary visualization with key insights")

# Additional detailed analysis - create a correlation matrix for binary features
binary_corr = binary_data.T.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(binary_corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1,
            linewidths=.5, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Between Binary Features', fontsize=16)
plt.tight_layout()
plt.savefig('feature_correlation.png', dpi=300, bbox_inches='tight')
plt.close()

print("4. feature_correlation.png - Correlation matrix showing relationships between binary features")

# Create an enhanced version to better highlight the most popular combinations
# This will help identify common patterns in virtual tutor design

# First, create a table counting language models by academic discipline
plt.figure(figsize=(12, 8))
model_disc = pd.crosstab(
    df[df['Questions'] == "Which language model is implemented?"].iloc[0, 1:],
    df[df['Questions'] == "Academic disciplines supported by the tutor?"].iloc[0, 1:]
)
sns.heatmap(model_disc, cmap="YlOrRd", annot=True, fmt="d", linewidths=.5)
plt.title('Language Models Used by Academic Discipline', fontsize=14)
plt.xlabel('Academic Discipline')
plt.ylabel('Language Model')
plt.tight_layout()
plt.savefig('model_by_discipline.png', dpi=300, bbox_inches='tight')
plt.close()

print("5. model_by_discipline.png - Distribution of language models across academic disciplines")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load the data
df = pd.read_csv('content_matrix.csv')

# Clean up by removing empty rows and columns
df = df.iloc[:10, :24]  # Only take the first 10 rows and 24 columns

# Dictionary to map full questions to shorter labels
question_labels = {
    "Does the system use Retrieval-Augmented Generation (RAG)?": "RAG",
    "Does the system support multimodal content?": "Multimodal",
    "Is student learning motivation explicitly addressed in the design?": "Motivation factor",
    "Assessment capabilities?": "Assessments",
    "Empirical evidence provided?": "Evidence",
    "Which language model is implemented?": "LLM",
    "Primary target audience of the virtual tutor?": "Target Audience",
    "Academic disciplines supported by the tutor?": "Disciplines",
    "Integration with learning management systems?": "Integration with LMS",
    "Implementation status?": "Implementation Status"
}

# For binary questions (0/1), we'll create a heatmap
binary_questions = [
    "Does the system use Retrieval-Augmented Generation (RAG)?",
    "Does the system support multimodal content?",
    "Is student learning motivation explicitly addressed in the design?",
    "Assessment capabilities?",
    "Empirical evidence provided?"
]

# For categorical questions, we'll create bar charts
categorical_questions = [
    "Which language model is implemented?",
    "Primary target audience of the virtual tutor?",
    "Academic disciplines supported by the tutor?",
    "Integration with learning management systems?",
    "Implementation status?"
]

# Create a figure for the heatmap of binary questions
plt.figure(figsize=(20, 6))
binary_data = df[df['Questions'].isin(binary_questions)].copy()

# Map the long question labels to shorter ones
binary_data = binary_data.copy()
binary_data['Questions'] = binary_data['Questions'].map(question_labels)
binary_data = binary_data.set_index('Questions')
binary_data = binary_data.astype(float)  # Convert to numeric

# Create heatmap
sns.heatmap(binary_data, cmap="YlGnBu", cbar_kws={'label': 'Feature Present'},
            linewidths=.5, linecolor='gray',
            annot=True, fmt=".0f", annot_kws={"size": 10})
plt.title('Binary Features Across Papers', fontsize=16)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.savefig('binary_features_heatmap.png', dpi=300, bbox_inches='tight')
plt.close()

# For categorical questions, let's generate frequency counts for the most common values
plt.figure(figsize=(20, 15))

for i, question in enumerate(categorical_questions):
    plt.subplot(3, 2, i+1)

    # Extract row data for this question
    row_data = df[df['Questions'] == question].iloc[0, 1:].value_counts()

    # Create bar chart
    colors = plt.cm.viridis(np.linspace(0, 0.9, len(row_data)))
    row_data.plot(kind='bar', color=colors)

    # Use shorter label for the title
    short_title = question_labels[question]
    plt.title(short_title, fontsize=14)
    plt.ylabel('Number of Papers', fontsize=10)
    plt.xlabel('Categories', fontsize=10)
    plt.xticks(rotation=45, ha='right', fontsize=9)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('categorical_features_barcharts.png', dpi=300, bbox_inches='tight')
plt.close()

# Create a comprehensive visual summary - simpler version focusing on key insights
plt.figure(figsize=(20, 10))

# Summarize language models
plt.subplot(2, 2, 1)
lm_data = df[df['Questions'] == "Which language model is implemented?"].iloc[0, 1:].value_counts()
lm_data.plot(kind='pie', autopct='%1.1f%%', colors=plt.cm.tab10.colors, startangle=90)
plt.title('Language Models Used', fontsize=14)
plt.ylabel('')

# Summarize implementation status
plt.subplot(2, 2, 2)
status_data = df[df['Questions'] == "Implementation status?"].iloc[0, 1:].value_counts()
status_data.plot(kind='pie', autopct='%1.1f%%', colors=plt.cm.Paired.colors, startangle=90)
plt.title('Implementation Status', fontsize=14)
plt.ylabel('')

# Summarize academic disciplines
plt.subplot(2, 2, 3)
discipline_data = df[df['Questions'] == "Academic disciplines supported by the tutor?"].iloc[0, 1:].value_counts()
discipline_data.plot(kind='bar', color=plt.cm.Set3.colors)
plt.title('Academic Disciplines', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Count of special features
plt.subplot(2, 2, 4)
features = []
counts = []

for q in binary_questions:
    feature_count = df[df['Questions'] == q].iloc[0, 1:].astype(float).sum()
    features.append(question_labels[q])
    counts.append(feature_count)

feature_df = pd.Series(counts, index=features)
feature_df.plot(kind='bar', color=plt.cm.tab20c.colors)
plt.title('Feature Adoption (out of 23 papers)', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylabel('Number of Papers')

plt.tight_layout()
plt.savefig('summary_visualization.png', dpi=300, bbox_inches='tight')
plt.close()

print("Visualizations generated:")
print("1. binary_features_heatmap.png - Heatmap showing binary features across papers")
print("2. categorical_features_barcharts.png - Bar charts showing distribution of categorical features")
print("3. summary_visualization.png - Summary visualization with key insights")

# Additional detailed analysis - create a correlation matrix for binary features
binary_corr = binary_data.T.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(binary_corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1,
            linewidths=.5, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Between Binary Features', fontsize=16)
plt.tight_layout()
plt.savefig('feature_correlation.png', dpi=300, bbox_inches='tight')
plt.close()

print("4. feature_correlation.png - Correlation matrix showing relationships between binary features")

# Create an enhanced version to better highlight the most popular combinations
# This will help identify common patterns in virtual tutor design

# First, create a table counting language models by academic discipline
plt.figure(figsize=(12, 8))
model_disc = pd.crosstab(
    df[df['Questions'] == "Which language model is implemented?"].iloc[0, 1:],
    df[df['Questions'] == "Academic disciplines supported by the tutor?"].iloc[0, 1:]
)
sns.heatmap(model_disc, cmap="YlOrRd", annot=True, fmt="d", linewidths=.5)
plt.title('LLMs by Academic Discipline', fontsize=14)
plt.xlabel('Academic Discipline')
plt.ylabel('Language Model')
plt.tight_layout()
plt.savefig('model_by_discipline.png', dpi=300, bbox_inches='tight')
plt.close()

print("5. model_by_discipline.png - Distribution of language models across academic disciplines")
```
